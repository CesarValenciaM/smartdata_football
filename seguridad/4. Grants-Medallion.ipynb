## Grants de seguridad

Ejemplos de GRANT, REVOKE, grupos y external locations para `football_dev`.


```python
GRANT USE CATALOG ON CATALOG football_dev TO `proyecto_smartdata1@outlook.com`;
GRANT USE SCHEMA ON SCHEMA football_dev.bronze TO `proyecto_smartdata1@outlook.com`;
GRANT USE SCHEMA ON SCHEMA football_dev.bronze TO `Desarrolladores`;
GRANT CREATE ON SCHEMA football_dev.bronze TO `Desarrolladores`;
```

```python
GRANT SELECT ON TABLE football_dev.bronze.matches TO `Analistas`;
GRANT SELECT ON TABLE football_dev.silver.matches_transformed TO `Analistas`;
GRANT SELECT ON TABLE football_dev.gold.season_stats TO `Admins_Smart`;
```

```python
SHOW GRANTS ON TABLE football_dev.bronze.matches;
```

```python
SHOW GRANTS ON SCHEMA football_dev.bronze;
```

```python
REVOKE USE SCHEMA ON SCHEMA football_dev.bronze FROM `proyecto_smartdata1@outlook.com`;
```
