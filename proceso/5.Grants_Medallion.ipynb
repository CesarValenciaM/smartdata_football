## 5. Grants_Medallion

Asigna permisos Unity Catalog sobre el catalogo de futbol.


```python
dbutils.widgets.removeAll()
```

```python
%sql
create widget text catalogo default "football_dev";
```

```python
%sql
GRANT USE CATALOG ON CATALOG $catalogo TO `cvalenciam@uni.pe`;
```

```python
%sql
GRANT USE SCHEMA ON SCHEMA $catalogo.bronze TO `cvalenciam@uni.pe`;
```

```python
%sql
GRANT SELECT ON TABLE $catalogo.bronze.leagues TO `cvalenciam@uni.pe`;
GRANT SELECT ON TABLE $catalogo.bronze.matches TO `cvalenciam@uni.pe`;
GRANT SELECT ON TABLE $catalogo.bronze.clubs TO `cvalenciam@uni.pe`;
GRANT SELECT ON TABLE $catalogo.silver.matches_transformed TO `cvalenciam@uni.pe`;
GRANT SELECT ON TABLE $catalogo.gold.season_stats TO `cvalenciam@uni.pe`;
```
