## 1. Preparacion_Ambiente

Crea las external locations, el catalogo y las tablas Delta vacias
de las capas bronze, silver y gold para el pipeline de futbol Big Five.


```python
dbutils.widgets.removeAll()
```

```python
%sql
create widget text storageName default "s3smartdata-bucket";
create widget text catalogo default "football_dev";
```

```python
%python
storageName = dbutils.widgets.get("storageName")
catalogo = dbutils.widgets.get("catalogo")
```

```python
%sql
DROP CATALOG IF EXISTS $catalogo CASCADE;
```

```python
%sql
CREATE CATALOG IF NOT EXISTS $catalogo
```

```python
%sql
DROP SCHEMA IF EXISTS $catalogo.raw;
DROP SCHEMA IF EXISTS $catalogo.bronze;
DROP SCHEMA IF EXISTS $catalogo.silver;
DROP SCHEMA IF EXISTS $catalogo.gold;
```

```python
%python
dbutils.fs.rm(f"s3://{storageName}/{catalogo}/raw", True)
dbutils.fs.rm(f"s3://{storageName}/{catalogo}/bronze", True)
dbutils.fs.rm(f"s3://{storageName}/{catalogo}/silver", True)
dbutils.fs.rm(f"s3://{storageName}/{catalogo}/gold", True)
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS $catalogo.raw;
CREATE SCHEMA IF NOT EXISTS $catalogo.bronze;
CREATE SCHEMA IF NOT EXISTS $catalogo.silver;
CREATE SCHEMA IF NOT EXISTS $catalogo.gold;
```

### Tablas Bronze


```python
%sql
CREATE TABLE IF NOT EXISTS $catalogo.bronze.leagues (
  league_id integer,
  league_ref string,
  name string,
  country string,
  country_code string,
  confederation string,
  tier integer,
  founded_year integer,
  ingestion_date timestamp
)
USING DELTA
LOCATION "s3://$storageName/$catalogo/bronze/leagues"
```

```python
%sql
CREATE TABLE IF NOT EXISTS $catalogo.bronze.matches (
  match_id integer,
  season_year integer,
  round integer,
  match_date date,
  league_id integer,
  home_club string,
  away_club string,
  home_goals integer,
  away_goals integer,
  ht_home_goals integer,
  ht_away_goals integer,
  ingestion_date timestamp
)
USING DELTA
PARTITIONED BY (season_year)
LOCATION "s3://$storageName/$catalogo/bronze/matches"
```

```python
%sql
CREATE TABLE IF NOT EXISTS $catalogo.bronze.clubs (
  club_id integer,
  club_ref string,
  name string,
  country string,
  city string,
  founded_year integer,
  league_id integer,
  ingestion_date timestamp
)
USING DELTA
LOCATION "s3://$storageName/$catalogo/bronze/clubs"
```

### Tablas Silver


```python
%sql
CREATE TABLE IF NOT EXISTS $catalogo.silver.matches_transformed (
  match_id integer,
  season_year integer,
  round integer,
  match_date date,
  league_id integer,
  league_name string,
  country string,
  home_club string,
  away_club string,
  home_club_ref string,
  away_club_ref string,
  home_city string,
  away_city string,
  home_goals integer,
  away_goals integer,
  total_goals integer,
  goal_diff integer,
  result_type string,
  goal_diff_category string,
  match_intensity string,
  is_classic string,
  season_age integer,
  ingestion_date timestamp
)
USING DELTA
LOCATION "s3://$storageName/$catalogo/silver/matches_transformed" 
```

### Tablas Golden


```python
%sql
CREATE TABLE IF NOT EXISTS $catalogo.gold.season_stats (
  season_year integer,
  country string,
  league_name string,
  conteo long,
  total_goals long,
  max_goals integer,
  min_goals integer,
  classic_count long,
  home_win_count long
)
USING DELTA
LOCATION "s3://$storageName/$catalogo/gold/season_stats" 
```
